```python
vendas_marca = df.groupby("Marca")["Valor Venda"].sum()

vendas_marca
```

```python
df["Produto"].nunique()
```

```python
quantidade_produto = df.groupby("Produto")["Quantidade"].sum()

quantidade_produto = quantidade_produto.sort_values(ascending=False)

quantidade_produto.head(10)
```

```python
top10_produtos = quantidade_produto.head(10)

top10_produtos.sort_values().plot(kind="barh")

plt.title("Top 10 Produtos por Quantidade Vendida")
plt.xlabel("Quantidade Vendida")
plt.ylabel("Produto")

plt.show()
```

```python
df["Custo Total"] = df["Custo Unitário"] * df["Quantidade"]
```

```python
df[["Custo Unitário", "Quantidade", "Custo Total"]].head()
```

```python
df["Lucro"] = df["Valor Venda"] - df["Custo Total"]
```

```python
df[["Valor Venda", "Custo Total", "Lucro"]].head()
```

```python
df["Lucro"].sum()
```

```python
lucro_total = df["Lucro"].sum()
vendas_total = df["Valor Venda"].sum()

margem = lucro_total / vendas_total

margem
```

```python
print(f"Margem de lucro: {margem:.2%}")
```

```python
lucro_ano = df.groupby(df["Data Venda"].dt.year)["Lucro"].sum()

lucro_ano
```

```python
lucro_ano.plot(kind="bar")

plt.title("Lucro por Ano")
plt.xlabel("Ano")
plt.ylabel("Lucro")

plt.show()
```

```python
lucro_marca = df.groupby("Marca")["Lucro"].sum()

lucro_marca = lucro_marca.sort_values(ascending=False)

lucro_marca
```

```python
lucro_marca.plot(kind="bar")

plt.title("Lucro por Marca")
plt.xlabel("Marca")
plt.ylabel("Lucro")

plt.show()
```

```python
df["Tempo de Envio"] = df["Data Envio"] - df["Data Venda"]
```

```python
df["Tempo de Envio"].head()
```

```python
df["Tempo de Envio"] = (df["Data Envio"] - df["Data Venda"]).dt.days
```

```python
df["Tempo de Envio"].head()
```

```python
df["Tempo de Envio"].mean()
```

```python
df["Tempo de Envio"].min()
```

```python
df["Tempo de Envio"].max()
```

```python
df["Tempo de Envio"].plot(kind="hist")

plt.title("Distribuição do Tempo de Envio")
plt.xlabel("Dias para envio")
plt.ylabel("Quantidade de vendas")

plt.show()
```
